```python
import pandas as pd
import folium
from folium.plugins import HeatMap
from IPython.display import display

# 1. Load data
url = 'https://docs.google.com/spreadsheets/d/1j6KBQp3XMkOnutERKrkHIF3eY3tqU3ViXo1yU68tGZk/export?format=csv&gid=1291050583'
df = pd.read_csv(url, na_values=['<Null>'], low_memory=False)
df = df.dropna()
df['Age'] = df['Age'].astype(int)
```

```python
# 2. Fix coordinates: swap inverted lat/lon and drop (0,0) points
mask = (df['latitude'] < 0) & (df['longitude'] > 0)
df.loc[mask, ['latitude', 'longitude']] = df.loc[mask, ['longitude', 'latitude']].values
df = df[df['latitude'].between(33, 37) & df['longitude'].between(-85, -75)]

# 3. Filter records after 2020
df['Date_of_Arrest'] = pd.to_datetime(df['Date_of_Arrest'])
df = df[df['Date_of_Arrest'].dt.year > 2020]

# 4. Clean and map categories
df.Gender = df.Gender.replace('M', 'Male')
df.Gender = df.Gender.replace('F', 'Female')

df.Drugs_or_Alcohol_Present = df.Drugs_or_Alcohol_Present.replace('Y', 'Yes')
df.Drugs_or_Alcohol_Present = df.Drugs_or_Alcohol_Present.replace('N', 'No')
df.Drugs_or_Alcohol_Present = df.Drugs_or_Alcohol_Present.replace('U', 'Unknown')

df.Ethnicity = df.Ethnicity.replace('H', 'Hispanic or Latino')
df.Ethnicity = df.Ethnicity.replace('N', 'Not Hispanic or Latino')

df.Race = df.Race.replace('B', 'Black')
df.Race = df.Race.replace('W', 'White')
df.Race = df.Race.replace('A', 'Asian')
df.Race = df.Race.replace('I', 'American Indian')

# Filter out unneeded categories
df = df[df['Race'] != 'O']
df = df[df['Race'] != 'H']
df = df[df['Ethnicity'] != 'U']
df = df.reset_index(drop=True)

df.head()
```

```python
# 5. Map Visualization Function
def arrest_map(col, param1, color1, color2):

    m = folium.Map(location=[df['latitude'].mean(), df['longitude'].mean()], tiles=None, zoom_start = 13.5)

    # OpenStreetMap
    folium.TileLayer(
        tiles="https://tile.openstreetmap.org/{z}/{x}/{y}.png",
        attr='&copy; <a href="https://www.openstreetmap.org/copyright">OpenStreetMap</a> contributors',
        referrerPolicy="no-referrer-when-downgrade",
        show=True,
        control=False
    ).add_to(m)

    fg = folium.FeatureGroup(name='Information/details of Person Arrested', show=True)
    for i in range(len(df)):
        if df[col].iloc[i] == param1:
            color = color1
        else:
            color = color2
        folium.Circle(
            location=[df.iloc[i]['latitude'], df.iloc[i]['longitude']],
            tooltip="<strong>Info: </strong>" + col + ' - ' + str(df.iloc[i][col])
            + "<br><strong>Primary Charge: </strong>" + str(df.iloc[i]['Primary_Charge'])
            + "<br><strong>Age: </strong>" + str(df.iloc[i]['Age']),
            radius=float(df.iloc[i]['Age']),
            stroke=False,
            fill=True,
            color=color,
            fill_opacity=0.5,
            control=True
        ).add_to(fg)
    fg.add_to(m)

    fgp = folium.FeatureGroup(name='Heatmap', show=True)
    heat_data = [[row['latitude'], row['longitude']] for index, row in df[df[col] == param1].iterrows()]
    HeatMap(heat_data).add_to(fgp)
    fgp.add_to(m)

    folium.LayerControl(position='topright', collapsed=False).add_to(m)
    display(m)
```

```python
# Example Call 1:
arrest_map('Race', 'White', 'cyan', 'black')
```

```python
# Example Call 2:
arrest_map('Drugs_or_Alcohol_Present', 'Yes', 'red', 'green')
```
